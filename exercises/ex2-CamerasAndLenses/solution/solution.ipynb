### Exercise 1

Explain how to calculate the angle $\theta$ when $a$ and $b$ is given
in the figure below. Calculate $\theta$ (in degrees) when
$a = 10$ and $b=3$ using the function `math.atan2()`. Remember to import `math` and find out what `atan2` does.

![Angle in triangel](figures/ArcTangens.png)

```python
# 1. 
import math

# Given values
a = 10  # horizontal side
b = 3   # vertical side

# Calculate angle θ using atan2
# atan2(y, x) where y is opposite side (b) and x is adjacent side (a)
theta_radians = math.atan2(b, a)

# Convert to degrees
theta_degrees = math.degrees(theta_radians)

print(f"Angle θ = {theta_degrees:.2f} degrees")
print(f"Angle θ = {theta_radians:.4f} radians")

# Explanation: math.atan2(y, x) calculates the angle from the positive x-axis
# to the point (x, y), handling all quadrants correctly.
```

```python
# 2. 
def camera_b_distance(f, g):
    """
    camera_b_distance returns the distance (b) where the CCD should be placed
    when the object distance (g) and the focal length (f) are given
    :param f: Focal length
    :param g: Object distance
    :return: b, the distance where the CCD should be placed
    """
    b = (f * g) / (g - f)
    return b
```

### Exercise 3

Thomas is 1.8 meters tall and standing 5 meters from a camera. The
cameras focal length is 5 mm. The CCD in the camera can be seen in
the figure below. It is a 1/2" (inches) CCD chip and the
image formed by the CCD is 640x480 pixels in a (x,y) coordinate system.

![CCD chip](../figures/CCDChip.png)

It is easiest to start by drawing the scene. The scene should
contain the optical axis, the optical center, the lens, the focal
point, the CCD chip, and Thomas. Do it on paper or even better in a
drawing program.

```python
# 3.1 A focused image of Thomas is formed inside the camera. At which distance from the lens?

# Given values
f = 5  # focal length in mm
g = 5000  # object distance in mm (Thomas is 5 meters away)

# Use the camera_b_distance function from exercise 2
b = camera_b_distance(f, g)

print(f"Focal length (f): {f} mm")
print(f"Object distance (g): {g} mm")
print(f"Image distance (b): {b:.4f} mm")

# The focused image of Thomas is formed at {b:.4f} mm from the lens
```

```python
# 3.2 How tall (in mm) will Thomas be on the CCD-chip?

# Given values
h = 1800  # height of Thomas in mm

# Calculate the height of Thomas on the CCD chip using similar triangles
h_image = (b / g) * h
print(f"Height of Thomas on CCD chip: {h_image:.4f} mm")
```

```python
# 3.3 What is the size of a single pixel on the CCD chip? (in mm)?
# CCD chip dimensions
ccd_width_mm = 6.4  # width in mm
ccd_height_mm = 4.8  # height in mm

# Calculate the size of a single pixel
num_pixels_width = 640  # number of pixels in width
num_pixels_height = 480  # number of pixels in height
pixel_width_mm = ccd_width_mm / num_pixels_width
pixel_height_mm = ccd_height_mm / num_pixels_height
print(f"Size of a single pixel on CCD chip: {pixel_width_mm:.4f} mm (width) x {pixel_height_mm:.4f} mm (height)")
```

```python
# 3.4 How many pixels tall will Thomas be on the CCD-chip?
pixels_tall = h_image / pixel_height_mm
print(f"Thomas will be approximately {pixels_tall:.2f} pixels tall on the CCD-chip")
```

```python
# 3.5 What is the horizontal field-of-view (in degrees)?

# Calculate horizontal field-of-view
# FOV = 2 * arctan((CCD_width/2) / focal_length)

half_angle_radians = math.atan(ccd_width_mm / (2 * f))
fov_radians = 2 * half_angle_radians
fov_degrees = math.degrees(fov_radians)

print(f"Horizontal field-of-view: {fov_degrees:.2f} degrees")

# Alternative calculation using atan2
fov_degrees_alt = 2 * math.degrees(math.atan2(ccd_width_mm/2, f))
print(f"Horizontal field-of-view (using atan2): {fov_degrees_alt:.2f} degrees")
```

```python
# 3.6 What is the vertical field-of-view (in degrees)?
# Calculate vertical field-of-view
# FOV = 2 * arctan((CCD_height/2) / focal_length)
half_angle_v_radians = math.atan(ccd_height_mm / (2 * f))
fov_v_radians = 2 * half_angle_v_radians
fov_v_degrees = math.degrees(fov_v_radians)
print(f"Vertical field-of-view: {fov_v_degrees:.2f} degrees")
```
